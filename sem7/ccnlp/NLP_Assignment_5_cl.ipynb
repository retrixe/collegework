### 5. Preprocessing

````python
print("\n" + "=" * 80)
print("PREPROCESSING")
print("=" * 80)

stop_words = set(stopwords.words("english"))
lemmatizer = WordNetLemmatizer()


def tokenize_sentences(text):
    """Split text into sentences."""
    return sent_tokenize(text)


def tokenize_words(text):
    """Tokenize text into words."""
    return word_tokenize(text)


def clean_words(text, lemmatize=True):
    """
    Lowercase, tokenize, remove punctuation and stopwords,
    and optionally lemmatize.
    """
    tokens = word_tokenize(text.lower())

    cleaned = []

    for token in tokens:
        if not token.isalpha():
            continue

        if token in stop_words:
            continue

        if lemmatize:
            token = lemmatizer.lemmatize(token)

        cleaned.append(token)

    return cleaned


def calculate_word_frequencies(text):
    """Calculate normalized word frequencies."""
    words = clean_words(text)

    frequencies = Counter(words)

    if not frequencies:
        return {}

    max_frequency = max(frequencies.values())

    normalized = {
        word: count / max_frequency
        for word, count in frequencies.items()
    }

    return normalized


def preprocess_document(text):
    """Run the complete preprocessing pipeline."""
    sentences = tokenize_sentences(text)
    words = tokenize_words(text)
    cleaned = clean_words(text)

    return {
        "sentences": sentences,
        "words": words,
        "cleaned_words": cleaned,
        "frequencies": Counter(cleaned)
    }


processed_data = preprocess_document(document)

print(f"Number of sentences: {len(processed_data['sentences'])}")
print(f"Number of tokens: {len(processed_data['words'])}")
print(f"Number of meaningful words: {len(processed_data['cleaned_words'])}")

print("\nMost frequent meaningful words:")
print(processed_data["frequencies"].most_common(15))

print("""
Why preprocessing is important:
- Sentence tokenization identifies candidate sentences.
- Word tokenization breaks sentences into analyzable units.
- Lowercasing treats words such as AI and ai consistently.
- Stopword removal reduces the influence of common grammatical words.
- Punctuation removal focuses scoring on meaningful terms.
- Lemmatization combines related word forms.
- Word frequencies provide useful signals for extractive methods.
""")
````

### 6. General Helper Functions

````python
def safe_sentence_limit(text, summary_ratio=0.20, num_sentences=5):
    """
    Determine the maximum number of sentences.

    Both summary_ratio and num_sentences are used:
    - summary_ratio limits the percentage of source sentences.
    - num_sentences provides a direct upper bound.
    """
    sentences = sent_tokenize(text)

    if not sentences:
        return 0

    ratio_limit = max(1, math.ceil(len(sentences) * summary_ratio))
    return max(1, min(num_sentences, ratio_limit, len(sentences)))


def select_top_sentences(sentences, scores, limit):
    """Select top-scoring sentences while preserving original document order."""
    if not sentences or limit <= 0:
        return ""

    ranked_indices = sorted(
        range(len(sentences)),
        key=lambda i: (-scores[i], i)
    )

    selected = sorted(ranked_indices[:limit])

    return " ".join(sentences[i] for i in selected)


def count_words(text):
    """Count alphabetic words."""
    return len([
        token for token in word_tokenize(text)
        if token.isalpha()
    ])


def calculate_compression_ratio(summary, original):
    """
    Compression ratio = summary word count / original word count.
    Lower values indicate stronger compression.
    """
    original_words = count_words(original)
    summary_words = count_words(summary)

    if original_words == 0:
        return 0.0

    return summary_words / original_words


def calculate_length_score(sentence):
    """
    Content/length score.
    Very short sentences are usually less informative.
    Extremely long sentences are also normalized.
    """
    words = clean_words(sentence)

    if not words:
        return 0.0

    length = len(words)

    # A soft preference for medium-length informative sentences.
    score = min(length / 20.0, 1.0)

    return score
````

### 7. Summarizer 1 - Frequency Based

````python
print("\n" + "=" * 80)
print("SUMMARIZER 1 - FREQUENCY BASED")
print("=" * 80)


def frequency_summarizer(text, summary_ratio=0.20, num_sentences=5):
    """
    Frequency-based extractive summarization.

    Each sentence receives a score based on the normalized frequency
    of its meaningful words.
    """
    sentences = sent_tokenize(text)

    if not sentences:
        return ""

    limit = safe_sentence_limit(
        text,
        summary_ratio,
        num_sentences
    )

    frequencies = calculate_word_frequencies(text)

    sentence_scores = []

    for sentence in sentences:
        words = clean_words(sentence)

        if not words:
            sentence_scores.append(0.0)
            continue

        score = sum(
            frequencies.get(word, 0.0)
            for word in words
        ) / len(words)

        sentence_scores.append(score)

    return select_top_sentences(
        sentences,
        sentence_scores,
        limit
    )
````

### 8. Summarizer 2 - Tf-Idf

````python
print("\n" + "=" * 80)
print("SUMMARIZER 2 - TF-IDF")
print("=" * 80)


def tfidf_summarizer(text, summary_ratio=0.20, num_sentences=5):
    """
    TF-IDF extractive summarizer.

    Each sentence is represented using TF-IDF.
    Its score is the average TF-IDF value of its meaningful words.
    """
    sentences = sent_tokenize(text)

    if not sentences:
        return ""

    limit = safe_sentence_limit(
        text,
        summary_ratio,
        num_sentences
    )

    if len(sentences) == 1:
        return sentences[0]

    vectorizer = TfidfVectorizer(
        stop_words="english",
        token_pattern=r"(?u)\b[a-zA-Z][a-zA-Z]+\b"
    )

    try:
        matrix = vectorizer.fit_transform(sentences)
    except ValueError:
        return select_top_sentences(
            sentences,
            [1.0] * len(sentences),
            limit
        )

    sentence_scores = np.asarray(matrix.mean(axis=1)).ravel()

    return select_top_sentences(
        sentences,
        sentence_scores,
        limit
    )
````

### 9. Summarizer 3 - Textrank

````python
print("\n" + "=" * 80)
print("SUMMARIZER 3 - TEXTRANK")
print("=" * 80)


def build_sentence_similarity_matrix(sentences):
    """
    Build a sentence similarity graph using TF-IDF and cosine similarity.
    """
    if len(sentences) == 1:
        return np.array([[1.0]])

    vectorizer = TfidfVectorizer(
        stop_words="english",
        token_pattern=r"(?u)\b[a-zA-Z][a-zA-Z]+\b"
    )

    try:
        tfidf_matrix = vectorizer.fit_transform(sentences)
        similarity_matrix = cosine_similarity(tfidf_matrix)
    except ValueError:
        similarity_matrix = np.zeros(
            (len(sentences), len(sentences))
        )

    # No self-links.
    np.fill_diagonal(similarity_matrix, 0.0)

    return similarity_matrix


def pagerank_scores(
    similarity_matrix,
    damping=0.85,
    iterations=100,
    tolerance=1e-6
):
    """
    Simple PageRank implementation for the sentence graph.
    """
    n = similarity_matrix.shape[0]

    if n == 0:
        return np.array([])

    scores = np.ones(n) / n

    row_sums = similarity_matrix.sum(axis=1)

    for _ in range(iterations):
        new_scores = np.ones(n) * (1 - damping) / n

        for i in range(n):
            if row_sums[i] == 0:
                new_scores += damping * scores[i] / n
            else:
                for j in range(n):
                    if similarity_matrix[i, j] > 0:
                        new_scores[j] += (
                            damping
                            * scores[i]
                            * similarity_matrix[i, j]
                            / row_sums[i]
                        )

        if np.linalg.norm(new_scores - scores) < tolerance:
            scores = new_scores
            break

        scores = new_scores

    return scores


def textrank_summarizer(text, summary_ratio=0.20, num_sentences=5):
    """
    Simplified TextRank extractive summarizer.

    Steps:
    1. Represent sentences with TF-IDF.
    2. Calculate cosine similarity.
    3. Build a sentence graph.
    4. Apply PageRank.
    5. Select highest-ranked sentences.
    """
    sentences = sent_tokenize(text)

    if not sentences:
        return ""

    limit = safe_sentence_limit(
        text,
        summary_ratio,
        num_sentences
    )

    if len(sentences) == 1:
        return sentences[0]

    similarity_matrix = build_sentence_similarity_matrix(sentences)

    scores = pagerank_scores(similarity_matrix)

    return select_top_sentences(
        sentences,
        scores,
        limit
    )
````

### 10. Summarizer 4 - Position + Frequency Hybrid

````python
print("\n" + "=" * 80)
print("SUMMARIZER 4 - POSITION + FREQUENCY HYBRID")
print("=" * 80)


def hybrid_summarizer(
    text,
    summary_ratio=0.20,
    num_sentences=5,
    frequency_weight=0.7
):
    """
    Hybrid extractive summarizer.

    Three components:
    - Word frequency
    - Sentence position
    - Sentence length/content

    The only configurable hybrid weight is frequency_weight.
    position_weight = 1 - frequency_weight.

    A small fixed content component is included so that the final
    score also considers sentence informativeness/length.

    Formula:

    Final Score =
        0.85 * (
            frequency_weight * Frequency
            + position_weight * Position
        )
        + 0.15 * Content/Length
    """
    sentences = sent_tokenize(text)

    if not sentences:
        return ""

    limit = safe_sentence_limit(
        text,
        summary_ratio,
        num_sentences
    )

    frequency_weight = float(
        np.clip(frequency_weight, 0.0, 1.0)
    )

    position_weight = 1.0 - frequency_weight

    frequencies = calculate_word_frequencies(text)

    frequency_scores = []

    for sentence in sentences:
        words = clean_words(sentence)

        if not words:
            frequency_scores.append(0.0)
        else:
            score = sum(
                frequencies.get(word, 0.0)
                for word in words
            ) / len(words)

            frequency_scores.append(score)

    # Normalize frequency scores.
    max_frequency_score = max(frequency_scores) if frequency_scores else 1

    if max_frequency_score > 0:
        frequency_scores = [
            score / max_frequency_score
            for score in frequency_scores
        ]

    # Earlier sentences receive higher position scores.
    n = len(sentences)

    position_scores = [
        (n - i) / n
        for i in range(n)
    ]

    content_scores = [
        calculate_length_score(sentence)
        for sentence in sentences
    ]

    final_scores = []

    for freq, pos, content in zip(
        frequency_scores,
        position_scores,
        content_scores
    ):
        score = (
            0.85 * (
                frequency_weight * freq
                + position_weight * pos
            )
            + 0.15 * content
        )

        final_scores.append(score)

    return select_top_sentences(
        sentences,
        final_scores,
        limit
    )
````

### 13. Rouge Evaluation

````python
print("\n" + "=" * 80)
print("ROUGE EVALUATION")
print("=" * 80)


rouge = rouge_scorer.RougeScorer(
    ["rouge1", "rouge2", "rougeL"],
    use_stemmer=True
)


def calculate_rouge(summary, reference):
    """Calculate ROUGE precision, recall and F1."""
    scores = rouge.score(reference, summary)

    return {
        "ROUGE-1 Precision": scores["rouge1"].precision,
        "ROUGE-1 Recall": scores["rouge1"].recall,
        "ROUGE-1 F1": scores["rouge1"].fmeasure,

        "ROUGE-2 Precision": scores["rouge2"].precision,
        "ROUGE-2 Recall": scores["rouge2"].recall,
        "ROUGE-2 F1": scores["rouge2"].fmeasure,

        "ROUGE-L Precision": scores["rougeL"].precision,
        "ROUGE-L Recall": scores["rougeL"].recall,
        "ROUGE-L F1": scores["rougeL"].fmeasure
    }


rouge_results = []

for method, summary in summaries.items():

    scores = calculate_rouge(
        summary,
        reference_summary
    )

    rouge_results.append({
        "Method": method,
        **scores
    })


rouge_df = pd.DataFrame(rouge_results)

print("\nComplete ROUGE Results:")
display(
    rouge_df.style.format({
        column: "{:.3f}"
        for column in rouge_df.columns
        if column != "Method"
    })
)

print("""
ROUGE-1:
Measures overlap of individual words (unigrams).

ROUGE-2:
Measures overlap of two-word sequences (bigrams).

ROUGE-L:
Measures similarity based on the longest common subsequence.

Precision:
How much of the generated summary overlaps with the reference.

Recall:
How much of the reference information is recovered.

F1:
The harmonic mean of precision and recall.
""")
````

### 14. Additional Metrics - Content Coverage

````python
print("\n" + "=" * 80)
print("ADDITIONAL METRICS")
print("=" * 80)


def calculate_content_coverage(summary, reference, threshold=0.15):
    """
    Estimate sentence/content coverage.

    Method:
    1. Represent reference sentences and generated-summary sentences
       using TF-IDF.
    2. Calculate cosine similarity between every reference sentence
       and summary sentence.
    3. A reference sentence is considered represented if its best
       matching summary sentence has similarity >= threshold.
    4. Coverage = represented reference sentences / total reference
       sentences.

    This is an approximate content-coverage metric, not a semantic
    human judgment metric.
    """
    reference_sentences = sent_tokenize(reference)
    summary_sentences = sent_tokenize(summary)

    if not reference_sentences:
        return 0.0

    if not summary_sentences:
        return 0.0

    all_sentences = reference_sentences + summary_sentences

    vectorizer = TfidfVectorizer(
        stop_words="english",
        token_pattern=r"(?u)\b[a-zA-Z][a-zA-Z]+\b"
    )

    try:
        matrix = vectorizer.fit_transform(all_sentences)
    except ValueError:
        return 0.0

    reference_vectors = matrix[:len(reference_sentences)]
    summary_vectors = matrix[len(reference_sentences):]

    similarities = cosine_similarity(
        reference_vectors,
        summary_vectors
    )

    best_matches = similarities.max(axis=1)

    covered = np.sum(best_matches >= threshold)

    return float(covered / len(reference_sentences))


additional_results = []

for method, summary in summaries.items():

    coverage = calculate_content_coverage(
        summary,
        reference_summary
    )

    compression = calculate_compression_ratio(
        summary,
        document
    )

    additional_results.append({
        "Method": method,
        "Compression Ratio": compression,
        "Content Coverage": coverage
    })


additional_df = pd.DataFrame(additional_results)

print("""
Content Coverage:
For every reference-summary sentence, the system finds the most similar
sentence in the generated summary using TF-IDF cosine similarity.

If the best similarity reaches the threshold, that reference sentence is
considered covered.

Coverage = covered reference sentences / total reference sentences.

Limitation:
This is only an approximate lexical-content metric. It can miss semantic
similarity when different words express the same idea, and it can also
consider lexical similarity important even when the meaning is not identical.
Therefore it should be interpreted together with ROUGE and human judgment.
""")

display(
    additional_df.style.format({
        "Compression Ratio": "{:.3f}",
        "Content Coverage": "{:.3f}"
    })
)
````

### 16. Parameter Tuning Experiment

````python
print("\n" + "=" * 80)
print("PARAMETER TUNING EXPERIMENT")
print("=" * 80)

# Three parameter ranges requested by the project.
summary_ratios = [0.10, 0.20, 0.30]
num_sentences_values = [3, 5, 8]
frequency_weights = [0.30, 0.50, 0.70]

print("Summary ratios:", summary_ratios)
print("Number of sentences:", num_sentences_values)
print("Hybrid frequency weights:", frequency_weights)

"""
Experiment design:

Frequency, TF-IDF and TextRank:
    3 summary ratios x 3 sentence limits = 9 configurations per method.

Hybrid:
    3 summary ratios x 3 sentence limits x 3 frequency weights
    = 27 configurations.

Total = 54 configurations.

This is large enough to investigate parameter effects while remaining
manageable in Google Colab.
"""

experiment_results = []

for ratio in summary_ratios:

    for sentence_limit in num_sentences_values:

        # --------------------------------------------------------
        # Frequency
        # --------------------------------------------------------

        frequency_summary = frequency_summarizer(
            document,
            ratio,
            sentence_limit
        )

        frequency_scores = calculate_rouge(
            frequency_summary,
            reference_summary
        )

        experiment_results.append({
            "Method": "Frequency",
            "Summary Ratio": ratio,
            "Number of Sentences": sentence_limit,
            "Frequency Weight": np.nan,
            "ROUGE-1": frequency_scores["ROUGE-1 F1"],
            "ROUGE-2": frequency_scores["ROUGE-2 F1"],
            "ROUGE-L": frequency_scores["ROUGE-L F1"],
            "Compression Ratio": calculate_compression_ratio(
                frequency_summary,
                document
            )
        })

        # --------------------------------------------------------
        # TF-IDF
        # --------------------------------------------------------

        tfidf_summary = tfidf_summarizer(
            document,
            ratio,
            sentence_limit
        )

        tfidf_scores = calculate_rouge(
            tfidf_summary,
            reference_summary
        )

        experiment_results.append({
            "Method": "TF-IDF",
            "Summary Ratio": ratio,
            "Number of Sentences": sentence_limit,
            "Frequency Weight": np.nan,
            "ROUGE-1": tfidf_scores["ROUGE-1 F1"],
            "ROUGE-2": tfidf_scores["ROUGE-2 F1"],
            "ROUGE-L": tfidf_scores["ROUGE-L F1"],
            "Compression Ratio": calculate_compression_ratio(
                tfidf_summary,
                document
            )
        })

        # --------------------------------------------------------
        # TextRank
        # --------------------------------------------------------

        textrank_summary = textrank_summarizer(
            document,
            ratio,
            sentence_limit
        )

        textrank_scores = calculate_rouge(
            textrank_summary,
            reference_summary
        )

        experiment_results.append({
            "Method": "TextRank",
            "Summary Ratio": ratio,
            "Number of Sentences": sentence_limit,
            "Frequency Weight": np.nan,
            "ROUGE-1": textrank_scores["ROUGE-1 F1"],
            "ROUGE-2": textrank_scores["ROUGE-2 F1"],
            "ROUGE-L": textrank_scores["ROUGE-L F1"],
            "Compression Ratio": calculate_compression_ratio(
                textrank_summary,
                document
            )
        })

        # --------------------------------------------------------
        # Hybrid - vary all three parameters
        # --------------------------------------------------------

        for weight in frequency_weights:

            hybrid_summary = hybrid_summarizer(
                document,
                ratio,
                sentence_limit,
                weight
            )

            hybrid_scores = calculate_rouge(
                hybrid_summary,
                reference_summary
            )

            experiment_results.append({
                "Method": "Hybrid",
                "Summary Ratio": ratio,
                "Number of Sentences": sentence_limit,
                "Frequency Weight": weight,
                "ROUGE-1": hybrid_scores["ROUGE-1 F1"],
                "ROUGE-2": hybrid_scores["ROUGE-2 F1"],
                "ROUGE-L": hybrid_scores["ROUGE-L F1"],
                "Compression Ratio": calculate_compression_ratio(
                    hybrid_summary,
                    document
                )
            })


experiment_df = pd.DataFrame(experiment_results)

experiment_df["Overall Score"] = experiment_df[
    ["ROUGE-1", "ROUGE-2", "ROUGE-L"]
].mean(axis=1)

print(f"\nTotal experiment configurations: {len(experiment_df)}")

display(
    experiment_df.head(20).style.format({
        "Summary Ratio": "{:.2f}",
        "Frequency Weight": lambda x: "-" if pd.isna(x) else f"{x:.2f}",
        "ROUGE-1": "{:.3f}",
        "ROUGE-2": "{:.3f}",
        "ROUGE-L": "{:.3f}",
        "Compression Ratio": "{:.3f}",
        "Overall Score": "{:.3f}"
    })
)
````

### 7. Nltk Preprocessing

````python
stop_words = set(stopwords.words("english"))
lemmatizer = WordNetLemmatizer()


def preprocess_words(text):
    """
    NLTK preprocessing:
    - tokenization
    - lowercasing
    - punctuation removal
    - stopword removal
    - lemmatization
    """

    tokens = word_tokenize(text.lower())

    words = []

    for token in tokens:

        if not token.isalpha():
            continue

        if token in stop_words:
            continue

        token = lemmatizer.lemmatize(token)

        words.append(token)

    return words


def get_sentences(text):
    return sent_tokenize(text)


def get_word_frequencies(text):
    words = preprocess_words(text)

    frequencies = Counter(words)

    if not frequencies:
        return {}

    maximum = max(frequencies.values())

    return {
        word: frequency / maximum
        for word, frequency in frequencies.items()
    }


print("\nPreprocessing completed.")

print("\nMost common meaningful words:")
print(
    Counter(preprocess_words(document)).most_common(20)
)
````

### 9. General Sentence Selection Function

````python
def get_summary_limit(text, ratio, sentence_limit):

    sentences = sent_tokenize(text)

    if len(sentences) == 0:
        return 0

    ratio_limit = max(
        1,
        math.ceil(len(sentences) * ratio)
    )

    return min(
        ratio_limit,
        sentence_limit,
        len(sentences)
    )


def select_sentences(sentences, scores, limit):

    if not sentences:
        return ""

    ranked = sorted(
        range(len(sentences)),
        key=lambda i: (-scores[i], i)
    )

    selected_indices = sorted(
        ranked[:limit]
    )

    return " ".join(
        sentences[i]
        for i in selected_indices
    )
````

### 10. Type 1 - Extractive Summarization

````python
def extractive_summarizer(
    text,
    summary_ratio=0.20,
    num_sentences=5
):
    """
    Extractive summarization using NLTK word frequencies.

    Important sentences are selected directly from the original document.
    No new sentences are generated.
    """

    sentences = sent_tokenize(text)

    if not sentences:
        return ""

    limit = get_summary_limit(
        text,
        summary_ratio,
        num_sentences
    )

    frequencies = get_word_frequencies(text)

    scores = []

    for sentence in sentences:

        words = preprocess_words(sentence)

        if not words:
            scores.append(0)
            continue

        score = sum(
            frequencies.get(word, 0)
            for word in words
        )

        score = score / len(words)

        scores.append(score)

    return select_sentences(
        sentences,
        scores,
        limit
    )
````

### 11. Type 2 - Abstractive Summarization

````python
print("\nLoading T5 model for abstractive summarization...")

model_name = "t5-small"

tokenizer = T5Tokenizer.from_pretrained(
    model_name
)

model = T5ForConditionalGeneration.from_pretrained(
    model_name
)

print("T5 model loaded.")


def abstractive_summarizer(
    text,
    summary_ratio=0.20,
    num_sentences=5
):
    """
    Abstractive summarization using T5.

    Unlike extractive methods, T5 can generate new wording.
    """

    # T5 has input-length constraints.
    words = text.split()

    if len(words) > 500:
        text_for_model = " ".join(words[:500])
    else:
        text_for_model = text

    input_text = "summarize: " + text_for_model

    inputs = tokenizer(
        input_text,
        return_tensors="pt",
        max_length=512,
        truncation=True
    )

    summary_length = max(
        40,
        min(
            150,
            int(len(words) * summary_ratio)
        )
    )

    output = model.generate(
        **inputs,
        max_length=summary_length,
        min_length=min(40, summary_length),
        num_beams=4,
        length_penalty=1.0,
        early_stopping=True
    )

    result = tokenizer.decode(
        output[0],
        skip_special_tokens=True
    )

    return result
````

### 12. Type 3 - Generic Summarization

````python
def generic_summarizer(
    text,
    summary_ratio=0.20,
    num_sentences=5
):
    """
    Generic summarization.

    Produces a general-purpose summary of the document without
    requiring a specific user query.

    It combines:
    - word frequency
    - sentence position
    - sentence length
    """

    sentences = sent_tokenize(text)

    if not sentences:
        return ""

    limit = get_summary_limit(
        text,
        summary_ratio,
        num_sentences
    )

    frequencies = get_word_frequencies(text)

    scores = []

    n = len(sentences)

    for i, sentence in enumerate(sentences):

        words = preprocess_words(sentence)

        if not words:
            frequency_score = 0
        else:
            frequency_score = sum(
                frequencies.get(word, 0)
                for word in words
            ) / len(words)

        # Position score.
        position_score = (n - i) / n

        # Prefer informative medium-length sentences.
        word_count = len(words)

        if word_count == 0:
            length_score = 0
        else:
            length_score = min(
                word_count / 20,
                1.0
            )

        final_score = (
            0.60 * frequency_score
            + 0.25 * position_score
            + 0.15 * length_score
        )

        scores.append(final_score)

    return select_sentences(
        sentences,
        scores,
        limit
    )
````

### 13. Type 4 - Query-Based Summarization

````python
def query_based_summarizer(
    text,
    query,
    summary_ratio=0.20,
    num_sentences=5
):
    """
    Query-based extractive summarization.

    The summary focuses specifically on information related
    to the user's query.
    """

    sentences = sent_tokenize(text)

    if not sentences:
        return ""

    limit = get_summary_limit(
        text,
        summary_ratio,
        num_sentences
    )

    # Combine query and document sentences for TF-IDF.
    documents = [query] + sentences

    vectorizer = TfidfVectorizer(
        stop_words="english"
    )

    matrix = vectorizer.fit_transform(documents)

    query_vector = matrix[0]

    sentence_vectors = matrix[1:]

    similarities = cosine_similarity(
        query_vector,
        sentence_vectors
    ).flatten()

    # Add a small frequency component.
    frequencies = get_word_frequencies(text)

    final_scores = []

    for i, sentence in enumerate(sentences):

        words = preprocess_words(sentence)

        if words:
            frequency_score = sum(
                frequencies.get(word, 0)
                for word in words
            ) / len(words)
        else:
            frequency_score = 0

        final_score = (
            0.80 * similarities[i]
            + 0.20 * frequency_score
        )

        final_scores.append(final_score)

    return select_sentences(
        sentences,
        final_scores,
        limit
    )
````

### 18. Content Coverage

````python
def content_coverage(
    summary,
    reference,
    threshold=0.15
):
    """
    Approximate content coverage.

    Each reference sentence is compared with generated-summary
    sentences using TF-IDF cosine similarity.

    If its best similarity >= threshold,
    the reference sentence is considered covered.
    """

    reference_sentences = sent_tokenize(
        reference
    )

    summary_sentences = sent_tokenize(
        summary
    )

    if not reference_sentences:
        return 0

    if not summary_sentences:
        return 0

    combined = (
        reference_sentences
        + summary_sentences
    )

    vectorizer = TfidfVectorizer(
        stop_words="english"
    )

    try:

        matrix = vectorizer.fit_transform(
            combined
        )

    except ValueError:

        return 0

    reference_matrix = matrix[
        :len(reference_sentences)
    ]

    summary_matrix = matrix[
        len(reference_sentences):
    ]

    similarity = cosine_similarity(
        reference_matrix,
        summary_matrix
    )

    best_matches = similarity.max(
        axis=1
    )

    covered = np.sum(
        best_matches >= threshold
    )

    return covered / len(
        reference_sentences
    )


coverage_results = []

for method, summary in summaries.items():

    coverage_results.append({

        "Method": method,

        "Content Coverage":
            content_coverage(
                summary,
                reference_summary
            ),

        "Compression Ratio":
            compression_ratio(
                summary,
                document
            )
    })


coverage_df = pd.DataFrame(
    coverage_results
)

print("\n" + "=" * 80)
print("ADDITIONAL METRICS")
print("=" * 80)

display(
    coverage_df.style.format({
        "Content Coverage": "{:.3f}",
        "Compression Ratio": "{:.3f}"
    })
)
````

### 20. Parameter Tuning

````python
summary_ratios = [
    0.10,
    0.20,
    0.30
]

sentence_values = [
    3,
    5,
    8
]

frequency_weights = [
    0.30,
    0.50,
    0.70
]


print("\n" + "=" * 80)
print("PARAMETER TUNING")
print("=" * 80)

print("Summary ratios:", summary_ratios)
print("Sentence limits:", sentence_values)
print("Frequency weights:", frequency_weights)


tuning_results = []


# ------------------------------------------------
# Extractive + Generic
# ------------------------------------------------

for ratio in summary_ratios:

    for n_sentences in sentence_values:

        for method in [
            "Extractive",
            "Generic"
        ]:

            if method == "Extractive":

                summary = extractive_summarizer(
                    document,
                    ratio,
                    n_sentences
                )

            else:

                summary = generic_summarizer(
                    document,
                    ratio,
                    n_sentences
                )

            scores = evaluate_rouge(
                summary,
                reference_summary
            )

            tuning_results.append({

                "Method": method,

                "Summary Ratio": ratio,

                "Number of Sentences":
                    n_sentences,

                "Frequency Weight": np.nan,

                "ROUGE-1":
                    scores["ROUGE-1 F1"],

                "ROUGE-2":
                    scores["ROUGE-2 F1"],

                "ROUGE-L":
                    scores["ROUGE-L F1"],

                "Compression Ratio":
                    compression_ratio(
                        summary,
                        document
                    )
            })


# ------------------------------------------------
# Query-Based
# ------------------------------------------------

for ratio in summary_ratios:

    for n_sentences in sentence_values:

        summary = query_based_summarizer(
            document,
            query,
            ratio,
            n_sentences
        )

        scores = evaluate_rouge(
            summary,
            reference_summary
        )

        tuning_results.append({

            "Method": "Query-Based",

            "Summary Ratio": ratio,

            "Number of Sentences":
                n_sentences,

            "Frequency Weight": np.nan,

            "ROUGE-1":
                scores["ROUGE-1 F1"],

            "ROUGE-2":
                scores["ROUGE-2 F1"],

            "ROUGE-L":
                scores["ROUGE-L F1"],

            "Compression Ratio":
                compression_ratio(
                    summary,
                    document
                )
        })


# ------------------------------------------------
# Abstractive
# ------------------------------------------------

for ratio in summary_ratios:

    for n_sentences in sentence_values:

        summary = abstractive_summarizer(
            document,
            ratio,
            n_sentences
        )

        scores = evaluate_rouge(
            summary,
            reference_summary
        )

        tuning_results.append({

            "Method": "Abstractive",

            "Summary Ratio": ratio,

            "Number of Sentences":
                n_sentences,

            "Frequency Weight": np.nan,

            "ROUGE-1":
                scores["ROUGE-1 F1"],

            "ROUGE-2":
                scores["ROUGE-2 F1"],

            "ROUGE-L":
                scores["ROUGE-L F1"],

            "Compression Ratio":
                compression_ratio(
                    summary,
                    document
                )
        })


# ------------------------------------------------
# Query / Hybrid frequency-weight style experiment
# ------------------------------------------------
# The third parameter is meaningful mainly to the
# frequency/position hybrid component. We therefore
# demonstrate it separately using the generic method.

def hybrid_generic_summarizer(
    text,
    ratio,
    n_sentences,
    frequency_weight
):

    sentences = sent_tokenize(text)

    if not sentences:
        return ""

    limit = get_summary_limit(
        text,
        ratio,
        n_sentences
    )

    frequencies = get_word_frequencies(text)

    n = len(sentences)

    scores = []

    for i, sentence in enumerate(sentences):

        words = preprocess_words(
            sentence
        )

        if words:

            frequency_score = sum(
                frequencies.get(
                    word,
                    0
                )
                for word in words
            ) / len(words)

        else:

            frequency_score = 0

        position_score = (
            n - i
        ) / n

        final_score = (
            frequency_weight
            * frequency_score
            +
            (1 - frequency_weight)
            * position_score
        )

        scores.append(
            final_score
        )

    return select_sentences(
        sentences,
        scores,
        limit
    )


for ratio in summary_ratios:

    for n_sentences in sentence_values:

        for weight in frequency_weights:

            summary = hybrid_generic_summarizer(
                document,
                ratio,
                n_sentences,
                weight
            )

            scores = evaluate_rouge(
                summary,
                reference_summary
            )

            tuning_results.append({

                "Method": "Hybrid Generic",

                "Summary Ratio": ratio,

                "Number of Sentences":
                    n_sentences,

                "Frequency Weight":
                    weight,

                "ROUGE-1":
                    scores["ROUGE-1 F1"],

                "ROUGE-2":
                    scores["ROUGE-2 F1"],

                "ROUGE-L":
                    scores["ROUGE-L F1"],

                "Compression Ratio":
                    compression_ratio(
                        summary,
                        document
                    )
            })


tuning_df = pd.DataFrame(
    tuning_results
)

tuning_df["Overall Score"] = tuning_df[
    [
        "ROUGE-1",
        "ROUGE-2",
        "ROUGE-L"
    ]
].mean(axis=1)


print("\nTuning experiment completed.")

display(
    tuning_df.head(20).style.format({
        "Summary Ratio": "{:.2f}",
        "Frequency Weight":
            lambda x:
            "-"
            if pd.isna(x)
            else f"{x:.2f}",
        "ROUGE-1": "{:.3f}",
        "ROUGE-2": "{:.3f}",
        "ROUGE-L": "{:.3f}",
        "Compression Ratio": "{:.3f}",
        "Overall Score": "{:.3f}"
    })
)
````

````python
print("\n" + "=" * 80)
print("ORIGINAL DOCUMENT")
print("=" * 80)
print(document.strip())

stop_words = set(stopwords.words("english"))
lemmatizer = WordNetLemmatizer()


def preprocess_words(text):
    """
    NLTK preprocessing:
    1. Lowercase
    2. Word tokenization
    3. Punctuation removal
    4. Stopword removal
    5. Lemmatization
    """

    tokens = word_tokenize(text.lower())

    clean_words = []

    for token in tokens:

        if not token.isalpha():
            continue

        if token in stop_words:
            continue

        token = lemmatizer.lemmatize(token)

        clean_words.append(token)

    return clean_words


def get_sentences(text):
    return sent_tokenize(text)


def word_count(text):
    """
    Consistent word-count function used throughout the project.
    """

    return len([
        word
        for word in word_tokenize(text)
        if word.isalpha()
    ])


def get_word_frequencies(text):

    words = preprocess_words(text)

    frequencies = Counter(words)

    if not frequencies:
        return {}

    maximum = max(frequencies.values())

    return {
        word: count / maximum
        for word, count in frequencies.items()
    }


print("\n" + "=" * 80)
print("PREPROCESSING")
print("=" * 80)

print("Number of sentences:",
      len(get_sentences(document)))

print("Number of meaningful words:",
      len(preprocess_words(document)))

print("\nTop 20 meaningful words:")

print(
    Counter(
        preprocess_words(document)
    ).most_common(20)
)
````

### 9. Determine Summary Length

````python
def calculate_target_sentences(
    text,
    summary_ratio,
    num_sentences
):
    """
    Summary Ratio determines the target length.
    Number of Sentences acts as an upper limit.

    This allows both parameters to have meaningful effects.
    """

    sentences = sent_tokenize(text)

    if not sentences:
        return 0

    ratio_target = max(
        1,
        round(
            len(sentences) * summary_ratio
        )
    )

    target = min(
        ratio_target,
        num_sentences,
        len(sentences)
    )

    return max(1, target)


def select_top_sentences(
    sentences,
    scores,
    number_to_select
):

    if not sentences:
        return ""

    ranked_indices = sorted(
        range(len(sentences)),
        key=lambda i: (
            -scores[i],
            i
        )
    )

    selected_indices = sorted(
        ranked_indices[
            :number_to_select
        ]
    )

    return " ".join(
        sentences[i]
        for i in selected_indices
    )
````

### 10. Extractive Summarizer

````python
def extractive_summarizer(
    text,
    summary_ratio=0.20,
    num_sentences=5
):

    sentences = sent_tokenize(text)

    if not sentences:
        return ""

    target = calculate_target_sentences(
        text,
        summary_ratio,
        num_sentences
    )

    frequencies = get_word_frequencies(text)

    scores = []

    for sentence in sentences:

        words = preprocess_words(
            sentence
        )

        if not words:
            scores.append(0)
            continue

        score = sum(
            frequencies.get(
                word,
                0
            )
            for word in words
        )

        score /= len(words)

        scores.append(score)

    return select_top_sentences(
        sentences,
        scores,
        target
    )
````

### 11. Generic Summarizer

````python
def generic_summarizer(
    text,
    summary_ratio=0.20,
    num_sentences=5,
    frequency_weight=0.70
):
    """
    Generic summarization.

    It does NOT use a query.

    It combines:
    - word frequency
    - sentence position
    - sentence content/length

    Frequency Weight is configurable.
    """

    sentences = sent_tokenize(text)

    if not sentences:
        return ""

    target = calculate_target_sentences(
        text,
        summary_ratio,
        num_sentences
    )

    frequencies = get_word_frequencies(text)

    position_weight = (
        1 - frequency_weight
    )

    scores = []

    total_sentences = len(sentences)

    for index, sentence in enumerate(
        sentences
    ):

        words = preprocess_words(
            sentence
        )

        if words:

            frequency_score = (
                sum(
                    frequencies.get(
                        word,
                        0
                    )
                    for word in words
                )
                / len(words)
            )

        else:

            frequency_score = 0

        # Beginning sentences receive slightly higher scores.
        position_score = (
            total_sentences - index
        ) / total_sentences

        # Content/length normalization.
        if words:

            length_score = min(
                len(words) / 20,
                1.0
            )

        else:

            length_score = 0

        final_score = (
            frequency_weight
            * frequency_score
            +
            position_weight
            * position_score
            +
            0.10
            * length_score
        )

        scores.append(
            final_score
        )

    return select_top_sentences(
        sentences,
        scores,
        target
    )
````

### 12. Query-Based Summarizer

````python
def query_based_summarizer(
    text,
    query,
    summary_ratio=0.20,
    num_sentences=5
):

    sentences = sent_tokenize(text)

    if not sentences:
        return ""

    target = calculate_target_sentences(
        text,
        summary_ratio,
        num_sentences
    )

    documents = [
        query
    ] + sentences

    vectorizer = TfidfVectorizer(
        stop_words="english"
    )

    matrix = vectorizer.fit_transform(
        documents
    )

    query_vector = matrix[0]

    sentence_vectors = matrix[1:]

    similarities = cosine_similarity(
        query_vector,
        sentence_vectors
    ).flatten()

    frequencies = get_word_frequencies(
        text
    )

    scores = []

    for index, sentence in enumerate(
        sentences
    ):

        words = preprocess_words(
            sentence
        )

        if words:

            frequency_score = (
                sum(
                    frequencies.get(
                        word,
                        0
                    )
                    for word in words
                )
                / len(words)
            )

        else:

            frequency_score = 0

        final_score = (
            0.80
            * similarities[index]
            +
            0.20
            * frequency_score
        )

        scores.append(
            final_score
        )

    return select_top_sentences(
        sentences,
        scores,
        target
    )
````

### 14. T5 Chunking Function

````python
def split_into_chunks(
    text,
    max_words=250
):
    """
    Splits a document into manageable word chunks.
    This prevents the abstractive model from only seeing
    the first part of a long document.
    """

    words = text.split()

    chunks = []

    for i in range(
        0,
        len(words),
        max_words
    ):

        chunks.append(
            " ".join(
                words[
                    i:i + max_words
                ]
            )
        )

    return chunks


def generate_t5_summary(
    text,
    max_length=100,
    min_length=30
):

    input_text = (
        "summarize: "
        + text
    )

    inputs = tokenizer(
        input_text,
        return_tensors="pt",
        max_length=512,
        truncation=True
    )

    output = model.generate(
        **inputs,
        max_length=max_length,
        min_length=min(
            min_length,
            max_length - 1
        ),
        num_beams=4,
        length_penalty=1.0,
        no_repeat_ngram_size=3,
        early_stopping=True
    )

    return tokenizer.decode(
        output[0],
        skip_special_tokens=True
    )
````

### 15. Abstractive Summarizer

````python
def abstractive_summarizer(
    text,
    summary_ratio=0.20,
    num_sentences=5
):
    """
    Abstractive summarization using T5.

    The document is divided into chunks.
    Each chunk is summarized.
    The intermediate summaries are then combined
    and summarized again.
    """

    chunks = split_into_chunks(
        text,
        max_words=250
    )

    intermediate_summaries = []

    for chunk in chunks:

        chunk_summary = generate_t5_summary(
            chunk,
            max_length=80,
            min_length=20
        )

        if chunk_summary.strip():

            intermediate_summaries.append(
                chunk_summary
            )

    if not intermediate_summaries:

        return ""

    combined = " ".join(
        intermediate_summaries
    )

    final_word_target = max(
        40,
        min(
            130,
            int(
                word_count(text)
                * summary_ratio
            )
        )
    )

    # If combined text is already short,
    # use it directly.
    if word_count(combined) <= final_word_target:

        return combined

    final_summary = generate_t5_summary(
        combined,
        max_length=final_word_target,
        min_length=max(
            20,
            min(
                50,
                final_word_target // 2
            )
        )
    )

    return final_summary
````

### 21. Content Coverage

````python
def content_coverage(
    summary,
    reference,
    threshold=0.15
):
    """
    Content Coverage:

    1. Split reference summary into sentences.
    2. Split generated summary into sentences.
    3. Convert sentences to TF-IDF vectors.
    4. Calculate cosine similarity.
    5. For each reference sentence, find its best matching
       generated sentence.
    6. A reference sentence is considered covered if its
       similarity is >= threshold.

    Coverage =
        covered reference sentences /
        total reference sentences
    """

    reference_sentences = sent_tokenize(
        reference
    )

    summary_sentences = sent_tokenize(
        summary
    )

    if not reference_sentences:

        return 0

    if not summary_sentences:

        return 0

    combined = (
        reference_sentences
        +
        summary_sentences
    )

    vectorizer = TfidfVectorizer(
        stop_words="english"
    )

    try:

        matrix = vectorizer.fit_transform(
            combined
        )

    except ValueError:

        return 0

    reference_matrix = matrix[
        :len(reference_sentences)
    ]

    summary_matrix = matrix[
        len(reference_sentences):
    ]

    similarities = cosine_similarity(
        reference_matrix,
        summary_matrix
    )

    best_similarity = similarities.max(
        axis=1
    )

    covered = np.sum(
        best_similarity >= threshold
    )

    return (
        covered
        /
        len(reference_sentences)
    )


coverage_results = []

for method, summary in summaries.items():

    coverage_results.append({

        "Method":
            method,

        "Content Coverage":
            content_coverage(
                summary,
                reference_summary
            ),

        "Compression Ratio":
            compression_ratio(
                summary,
                document
            )
    })


coverage_df = pd.DataFrame(
    coverage_results
)


print("\n" + "=" * 80)
print("ADDITIONAL EVALUATION METRICS")
print("=" * 80)

display(
    coverage_df.style.format({
        "Content Coverage":
            "{:.3f}",

        "Compression Ratio":
            "{:.3f}"
    })
)
````

### 40. Parameter Effect Analysis

````python
ratio_scores = (
    tuning_df
    .groupby(
        "Summary Ratio"
    )[
        [
            "ROUGE-1",
            "ROUGE-2",
            "ROUGE-L"
        ]
    ]
    .mean()
)


ratio_scores["Mean ROUGE"] = (
    ratio_scores.mean(
        axis=1
    )
)


sentence_scores = (
    tuning_df
    .groupby(
        "Number of Sentences"
    )[
        [
            "ROUGE-1",
            "ROUGE-2",
            "ROUGE-L"
        ]
    ]
    .mean()
)


sentence_scores["Mean ROUGE"] = (
    sentence_scores.mean(
        axis=1
    )
)


weight_scores = (
    tuning_df[
        tuning_df["Method"] ==
        "Generic"
    ]
    .groupby(
        "Frequency Weight"
    )[
        [
            "ROUGE-1",
            "ROUGE-2",
            "ROUGE-L"
        ]
    ]
    .mean()
)


weight_scores["Mean ROUGE"] = (
    weight_scores.mean(
        axis=1
    )
)


print("\n" + "=" * 80)
print("PARAMETER EFFECT ANALYSIS")
print("=" * 80)

print("\nSummary Ratio Effect:")

display(
    ratio_scores.style.format(
        "{:.3f}"
    )
)

print("\nNumber of Sentences Effect:")

display(
    sentence_scores.style.format(
        "{:.3f}"
    )
)

print("\nFrequency Weight Effect:")

display(
    weight_scores.style.format(
        "{:.3f}"
    )
)
````

### 41. Determine Largest Parameter Effect

````python
ratio_effect_size = (
    ratio_scores[
        "Mean ROUGE"
    ].max()
    -
    ratio_scores[
        "Mean ROUGE"
    ].min()
)


sentence_effect_size = (
    sentence_scores[
        "Mean ROUGE"
    ].max()
    -
    sentence_scores[
        "Mean ROUGE"
    ].min()
)


weight_effect_size = (
    weight_scores[
        "Mean ROUGE"
    ].max()
    -
    weight_scores[
        "Mean ROUGE"
    ].min()
)


parameter_effects = {

    "Summary Ratio":
        ratio_effect_size,

    "Number of Sentences":
        sentence_effect_size,

    "Frequency Weight":
        weight_effect_size
}


largest_parameter = max(
    parameter_effects,
    key=parameter_effects.get
)
````

````python
print("\n" + "=" * 80)
print("PROJECT CONCLUSION")
print("=" * 80)

print(f"""
RESULTS
-------

The experiment compared four summarization approaches:

1. Extractive
2. Abstractive
3. Generic
4. Query-Based

Best-performing summarizer:
    {best_method}

Best Overall ROUGE Score:
    {best_score:.4f}

Lowest-performing summarizer:
    {worst_method}

Lowest Overall ROUGE Score:
    {worst_score:.4f}


PARAMETER ANALYSIS
------------------

The parameter showing the largest average effect on ROUGE was:

    {largest_parameter}


INTERPRETATION
--------------

Extractive summarization selects existing sentences from the document.
Its main advantage is that it preserves the original wording and does not
invent new information.

Abstractive summarization uses a T5 language model to generate new text.
It can combine ideas and produce more natural summaries, but generative
models may sometimes produce information that is not perfectly faithful
to the source.

Generic summarization creates a broad summary without requiring a query.
The implemented method combines word frequency, sentence position, and
content length.

Query-based summarization focuses specifically on a user's information
need. In this experiment the query was:

"{query}"

A query-based system can be useful when a document contains many topics
but the user is interested in only one particular topic.

SUMMARY LENGTH
--------------

A larger summary ratio or larger sentence limit generally allows more
information to be retained. This can increase ROUGE recall because more
reference information may appear in the generated summary.

However, increasing summary length also reduces compression and may
introduce redundant or less important sentences. Therefore, the best
summary is not necessarily the longest summary.

EVALUATION LIMITATIONS
----------------------

ROUGE measures overlap between the generated summary and reference summary.
Therefore, a summary may receive a lower ROUGE score even if it is useful,
coherent, and semantically correct.

The Content Coverage metric used here is also an approximate TF-IDF-based
measure and should not be interpreted as a perfect semantic evaluation.

The experiment uses only one dummy document and one manually written
reference summary. Therefore, the winning method cannot be considered
universally superior.

FUTURE WORK
-----------

Possible improvements include:

- BERT-based summarization
- BART
- PEGASUS
- Larger T5 models
- Transformer-based semantic similarity
- Abstractive summarization with larger datasets
- Multiple reference summaries
- Human evaluation
- Domain-specific datasets
- Multilingual summarization
- Redundancy detection
- Sentence embeddings
- Larger benchmark datasets
""")
````
