```python
#text preprocessing
import nltk
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer
nltk.download("stopwords")
nltk.download("wordnet")
raw_sentence = "The striped bats are hanging upside down and eating delicious fruits."
print(f"Original Text: '{raw_sentence}'\n")
lowercased = raw_sentence.lower()
print(f"1. Lowercased:      {lowercased}")

tokens = nltk.word_tokenize(lowercased)
print(f"2. Tokenized:       {tokens}")

english_stopwords = set(stopwords.words("english"))
filtered_tokens = [word for word in tokens if word not in english_stopwords]
print(f"3. Stopwords Removed: {filtered_tokens}")

stemmer = PorterStemmer()
stemmed_words = [stemmer.stem(word) for word in filtered_tokens]
print(f"4. Stemmed (Crude):   {stemmed_words}")

lemmatizer = WordNetLemmatizer()

lemmed_words = [
    lemmatizer.lemmatize(word, pos="v") if word == "eating" else lemmatizer.lemmatize(word)
    for word in filtered_tokens
]
print(f"5. Lemmatized (Smart): {lemmed_words}")

pos_tags = nltk.pos_tag(filtered_tokens)
print(f"6. POS Tagged:         {pos_tags}")
```

```python
import matplotlib.pyplot as plt
import networkx as nx
from nltk.corpus import wordnet as wn

# 1. Initialize our blank connect-the-dots map
G = nx.DiGraph()


# 2. Define a helper function to climb up the tree
def build_tree(synset):
    parents = synset.hypernyms()
    if parents:
        parent = parents[0]
        # Clean the names (e.g., 'bank.n.01' becomes 'bank')
        parent_name = parent.name().split(".")[0]
        child_name = synset.name().split(".")[0]

        # Draw a line from parent to child
        G.add_edge(parent_name, child_name)

        # Keep climbing up from the parent
        build_tree(parent)


# 3. Grab both meanings of bank
bank_meanings = wn.synsets("bank")
river_bank = bank_meanings[0]
financial_bank = bank_meanings[1]

# 4. Build the tree for BOTH meanings
build_tree(river_bank)
build_tree(financial_bank)

# 5. Draw it on the screen
plt.figure(figsize=(12, 6))
pos = nx.spring_layout(G, seed=42)
nx.draw(
    G,
    pos,
    with_labels=True,
    node_color="lightgreen",
    node_size=2500,
    font_size=10,
    font_weight="bold",
)
plt.title("How WordNet Splits the Meanings of 'Bank'")
plt.show()
```

```python
import matplotlib.pyplot as plt
import networkx as nx
from nltk.corpus import wordnet as wn

search_results = wn.synsets("car")

center_concept = search_results[0]
G_cat = nx.DiGraph() # Directed Graph (blueprint for arrows)

G_cat.add_node(center_concept.name().split(".")[0])

if center_concept.hypernyms():
    parent = center_concept.hypernyms()[0]
    parent_word = parent.name().split(".")[0]
    # Draw arrow: Parent -> Child (Motor_vehicle -> Car)
    G_cat.add_edge(parent_word, center_concept.name().split(".")[0])


if center_concept.hyponyms():
    child = center_concept.hyponyms()[0]
    child_word = child.name().split(".")[0]
    # Draw arrow: Parent -> Child (Car -> convertible)
    G_cat.add_edge(center_concept.name().split(".")[0], child_word)


# --- Part 3: PAINT THE CANVAS (Drawing) ---
plt.figure(figsize=(10, 5))
# pos = best positions for bubbles so they don't crash
pos_cat = nx.spring_layout(G_cat, seed=10)

# nx.draw defines the look: green bubbles, clean labels, standard size
nx.draw(G_cat, pos_cat, with_labels=True, node_color="lightgreen", node_size=3000, font_weight="bold", arrows=True)

plt.title("Map 1: Categorization (Is-A Relationship)\n[Parent <- center_concept -> Child]")
plt.show()
```

```python
import string
import matplotlib.pyplot as plt
import networkx as nx
import nltk
from nltk.corpus import wordnet as wn

# 1. Grab all noun lemmas from WordNet
all_noun_lemmas = list(wn.all_lemma_names(pos="n"))

# Take the top 10 words
top_10_words = all_noun_lemmas[:10]
print("--- TOP 10 WORDS FROM WORDNET ---")
print(top_10_words)

# Pick the very first word from the list
target_word = top_10_words[0]
print(f"\n==========================================")
print(f"SELECTED WORD FOR ANALYSIS: '{target_word}'")
print(f"==========================================")

# Get its first synset meaning
synsets = wn.synsets(target_word, pos="n")
if not synsets:
    print(f"No noun synset found for '{target_word}'")
else:
    concept = synsets[0]
    concept_id = concept.name()
    concept_word = concept.name().split(".")[0]

    print(f"WordNet Synset ID: {concept_id}")
    print(f"Definition: {concept.definition()}\n")

    # ==========================================
    # 1. SYNONYMS & LEMMAS & ANTONYMS EXTRACTION
    # ==========================================
    synonyms = []
    lemmas_list = []
    antonyms = []

    for lemma in concept.lemmas():
        clean_lemma_name = lemma.name().replace("_", " ")
        lemmas_list.append(clean_lemma_name)
        synonyms.append(clean_lemma_name) # Synonyms are the names of the lemmas

        # Check if this specific lemma has any antonyms
        if lemma.antonyms():
            for ant in lemma.antonyms():
                antonyms.append(ant.name().replace("_", " "))

    print("--- WordNet Text Extracted Data ---")
    print(f"Lemmas:    {lemmas_list}")
    print(f"Synonyms:  {synonyms}")
    print(f"Antonyms:  {antonyms if antonyms else 'None found in WordNet for this word'}\n")

    # ==========================================
    # VISUALIZATION 1: SYNONYMS & LEMMAS
    # ==========================================
    G_syn = nx.DiGraph()
    G_syn.add_node(concept_id)
    for syn in synonyms:
        G_syn.add_edge(syn, concept_id)

    plt.figure(figsize=(6, 3))
    nx.draw(G_syn, nx.spring_layout(G_syn, seed=42), with_labels=True, node_color="gold", node_size=2500, font_weight="bold", arrows=True)
    plt.title(f"1. Lemmas/Synonyms pointing to Concept ID '{concept_id}'")
    plt.show()

    # ==========================================
    # VISUALIZATION 2: ANTONYMS (Opposites)
    # ==========================================
    if antonyms:
        G_ant = nx.DiGraph()
        for ant in set(antonyms):
            G_ant.add_edge(concept_word, ant)

        plt.figure(figsize=(6, 3))
        nx.draw(G_ant, nx.spring_layout(G_ant, seed=42), with_labels=True, node_color="orange", node_size=2500, font_weight="bold", arrows=True)
        plt.title(f"2. Antonyms (Opposites of '{concept_word}')")
        plt.show()
    else:
        print(f"ℹ️ Skipping Antonym Map: No direct antonyms listed for '{concept_word}'.")

    # ==========================================
    # VISUALIZATION 3: HYPERNYM (Parent Category)
    # ==========================================
    G_hyper = nx.DiGraph()
    if concept.hypernyms():
        parent_word = concept.hypernyms()[0].name().split(".")[0]
        G_hyper.add_edge(parent_word, concept_word)

        plt.figure(figsize=(6, 3))
        nx.draw(G_hyper, nx.spring_layout(G_hyper, seed=42), with_labels=True, node_color="lightgreen", node_size=2500, font_weight="bold", arrows=True)
        plt.title("3. HYPERNYM (Parent Category)")
        plt.show()
    else:
        print(f"ℹ️ No Hypernyms found for '{concept_word}'.")

    # ==========================================
    # VISUALIZATION 4: HYPONYM (Specific Child Type)
    # ==========================================
    G_hypo = nx.DiGraph()
    if concept.hyponyms():
        child_word = concept.hyponyms()[0].name().split(".")[0]
        G_hypo.add_edge(concept_word, child_word)

        plt.figure(figsize=(6, 3))
        nx.draw(G_hypo, nx.spring_layout(G_hypo, seed=42), with_labels=True, node_color="lightblue", node_size=2500, font_weight="bold", arrows=True)
        plt.title("4. HYPONYM (Specific Child Type)")
        plt.show()
    else:
        print(f"ℹ️ No Hyponyms found for '{concept_word}'.")

    # ==========================================
    # VISUALIZATION 5: MERONYM (Physical Part)
    # ==========================================
    G_mero = nx.DiGraph()
    if concept.part_meronyms():
        part_word = concept.part_meronyms()[0].name().split(".")[0]
        G_mero.add_edge(concept_word, part_word)

        plt.figure(figsize=(6, 3))
        nx.draw(G_mero, nx.spring_layout(G_mero, seed=42), with_labels=True, node_color="lightcoral", node_size=2500, font_weight="bold", arrows=True)
        plt.title("5. MERONYM (Physical Part)")
        plt.show()
    else:
        print(f"ℹ️ Note: '{concept_word}' has no physical part meronyms (it might be abstract).")

    # ==========================================
    # VISUALIZATION 6: HOLONYM (Whole Container)
    # ==========================================
    G_holo = nx.DiGraph()
    if concept.part_holonyms():
        whole_word = concept.part_holonyms()[0].name().split(".")[0]
        G_holo.add_edge(concept_word, whole_word)

        plt.figure(figsize=(6, 3))
        nx.draw(G_holo, nx.spring_layout(G_holo, seed=42), with_labels=True, node_color="plum", node_size=2500, font_weight="bold", arrows=True)
        plt.title("6. HOLONYM (Whole Container)")
        plt.show()
    else:
        print(f"ℹ️ Note: '{concept_word}' has no part holonyms listed.")
```

```python
import string
import matplotlib.pyplot as plt
import networkx as nx
import nltk
from nltk.corpus import wordnet as wn

all_noun_lemmas = list(wn.all_lemma_names(pos="n"))

clean_words = []
for word in all_noun_lemmas:

    if word.isalpha() and len(word) > 1:
        clean_words.append(word)

top_10_words = clean_words[:10]
print("--- TASK: TOP 10 REAL WORDS FROM WORDNET ---")
print(top_10_words)

target_word = top_10_words[0]
print(f"\n==========================================")
print(f"SELECTED WORD FOR ANALYSIS: '{target_word}'")
print(f"==========================================")

synsets = wn.synsets(target_word, pos="n")
if not synsets:
    print(f"No noun synset found for '{target_word}'")
else:
    concept = synsets[0]
    concept_id = concept.name()
    concept_word = concept.name().split(".")[0]

    print(f"WordNet Synset ID: {concept_id}")
    print(f"Definition: {concept.definition()}\n")

    # ==========================================
    # 1. SYNONYMS & LEMMAS & ANTONYMS EXTRACTION
    # ==========================================
    synonyms = []
    lemmas_list = []
    antonyms = []

    for lemma in concept.lemmas():
        clean_lemma_name = lemma.name().replace("_", " ")
        lemmas_list.append(clean_lemma_name)
        synonyms.append(clean_lemma_name)

        if lemma.antonyms():
            for ant in lemma.antonyms():
                antonyms.append(ant.name().replace("_", " "))

    print("--- WordNet Text Extracted Data ---")
    print(f"Lemmas:    {lemmas_list}")
    print(f"Synonyms:  {synonyms}")
    print(f"Antonyms:  {antonyms if antonyms else 'None found in WordNet for this word'}\n")

    # ==========================================
    # VISUALIZATION 1: SYNONYMS & LEMMAS
    # ==========================================
    G_syn = nx.DiGraph()
    G_syn.add_node(concept_id)
    for syn in synonyms:
        G_syn.add_edge(syn, concept_id)

    plt.figure(figsize=(6, 3))
    nx.draw(G_syn, nx.spring_layout(G_syn, seed=42), with_labels=True, node_color="gold", node_size=2500, font_weight="bold", arrows=True)
    plt.title(f"1. Lemmas/Synonyms pointing to Concept ID '{concept_id}'")
    plt.show()

    # ==========================================
    # VISUALIZATION 2: ANTONYMS (Opposites)
    # ==========================================
    if antonyms:
        G_ant = nx.DiGraph()
        for ant in set(antonyms):
            G_ant.add_edge(concept_word, ant)

        plt.figure(figsize=(6, 3))
        nx.draw(G_ant, nx.spring_layout(G_ant, seed=42), with_labels=True, node_color="orange", node_size=2500, font_weight="bold", arrows=True)
        plt.title(f"2. Antonyms (Opposites of '{concept_word}')")
        plt.show()
    else:
        print(f"ℹ️ Skipping Antonym Map: No direct antonyms listed for '{concept_word}'.")

    # ==========================================
    # VISUALIZATION 3: HYPERNYM (Parent Category)
    # ==========================================
    G_hyper = nx.DiGraph()
    if concept.hypernyms():
        parent_word = concept.hypernyms()[0].name().split(".")[0]
        G_hyper.add_edge(parent_word, concept_word)

        plt.figure(figsize=(6, 3))
        nx.draw(G_hyper, nx.spring_layout(G_hyper, seed=42), with_labels=True, node_color="lightgreen", node_size=2500, font_weight="bold", arrows=True)
        plt.title("3. HYPERNYM (Parent Category)")
        plt.show()
    else:
        print(f"ℹ️ No Hypernyms found for '{concept_word}'.")

    # ==========================================
    # VISUALIZATION 4: HYPONYM (Specific Child Type)
    # ==========================================
    G_hypo = nx.DiGraph()
    if concept.hyponyms():
        child_word = concept.hyponyms()[0].name().split(".")[0]
        G_hypo.add_edge(concept_word, child_word)

        plt.figure(figsize=(6, 3))
        nx.draw(G_hypo, nx.spring_layout(G_hypo, seed=42), with_labels=True, node_color="lightblue", node_size=2500, font_weight="bold", arrows=True)
        plt.title("4. HYPONYM (Specific Child Type)")
        plt.show()
    else:
        print(f"ℹ️ No Hyponyms found for '{concept_word}'.")

    # ==========================================
    # VISUALIZATION 5: MERONYM (Physical Part)
    # ==========================================
    G_mero = nx.DiGraph()
    if concept.part_meronyms():
        part_word = concept.part_meronyms()[0].name().split(".")[0]
        G_mero.add_edge(concept_word, part_word)

        plt.figure(figsize=(6, 3))
        nx.draw(G_mero, nx.spring_layout(G_mero, seed=42), with_labels=True, node_color="lightcoral", node_size=2500, font_weight="bold", arrows=True)
        plt.title("5. MERONYM (Physical Part)")
        plt.show()
    else:
        print(f"ℹ️ Note: '{concept_word}' has no physical part meronyms (it might be abstract).")

    # ==========================================
    # VISUALIZATION 6: HOLONYM (Whole Container)
    # ==========================================
    G_holo = nx.DiGraph()
    if concept.part_holonyms():
        whole_word = concept.part_holonyms()[0].name().split(".")[0]
        G_holo.add_edge(concept_word, whole_word)

        plt.figure(figsize=(6, 3))
        nx.draw(G_holo, nx.spring_layout(G_holo, seed=42), with_labels=True, node_color="plum", node_size=2500, font_weight="bold", arrows=True)
        plt.title("6. HOLONYM (Whole Container)")
        plt.show()
    else:
        print(f"ℹ️ Note: '{concept_word}' has no part holonyms listed.")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import pandas as pd

# Set nice visual style
sns.set_theme(style="whitegrid")
```

```python
from nltk.corpus import gutenberg
from nltk.probability import FreqDist

# Get word frequencies for Hamlet (excluding simple punctuation/stopwords)
words = [w.lower() for w in gutenberg.words('shakespeare-hamlet.txt') if w.isalpha()]
fdist = FreqDist(words)

top_10 = fdist.most_common(10)
word_labels, counts = zip(*top_10)

plt.figure(figsize=(9, 4))
sns.barplot(x=list(counts), y=list(word_labels), palette="viridis")
plt.title("Top 10 Most Frequent Words in Hamlet", fontsize=12, fontweight="bold")
plt.xlabel("Frequency Count")
plt.ylabel("Words")
plt.show()
```

```python
plt.figure(figsize=(9, 4))
# Plot cumulative frequency curve for top 25 words
fdist.plot(25, cumulative=True, title="Cumulative Word Frequency Distribution (Top 25)")
plt.show()
```

```python
import nltk

sample_text = gutenberg.raw('shakespeare-hamlet.txt')[:1000]
tokens = nltk.word_tokenize(sample_text)
tagged = nltk.pos_tag(tokens)

# Categorize coarse POS tags
pos_categories = {'Nouns': 0, 'Verbs': 0, 'Adjectives': 0, 'Others': 0}
for _, tag in tagged:
    if tag.startswith('NN'):
        pos_categories['Nouns'] += 1
    elif tag.startswith('VB'):
        pos_categories['Verbs'] += 1
    elif tag.startswith('JJ'):
        pos_categories['Adjectives'] += 1
    elif tag.isalpha():
        pos_categories['Others'] += 1

labels = list(pos_categories.keys())
sizes = list(pos_categories.values())
colors = ['#ff9999','#66b3ff','#99ff99','#ffcc99']

plt.figure(figsize=(6, 6))
plt.pie(sizes, labels=labels, colors=colors, autopct='%1.1f%%', startangle=140, explode=(0.05, 0, 0, 0))
plt.title("POS Tag Distribution in Text Sample", fontsize=12, fontweight="bold")
plt.show()
```
